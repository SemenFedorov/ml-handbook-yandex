```python
import numpy as np

def most_frequent(nums):

    count_array = []

    for i in range(len(nums)):

        counter = 0

        for j in range(len(nums)):
            if nums[i] == nums[j]:
                counter += 1
        count_array.append(counter)

    target_num = max(count_array)

    for i in range(len(count_array)):
        if count_array[i] == target_num:
            target_i = i

    return nums[target_i]

#print(most_frequent(np.array([1, 1, 1, 3, 4, 5, 2, 2])))
```
