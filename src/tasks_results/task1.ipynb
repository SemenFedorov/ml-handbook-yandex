```python
import numpy as np 

def construct_matrix(first_array, second_array):

    length_first = len(first_array)
    length_second = len(second_array)

    return np.hstack((first_array.reshape(length_first, 1), second_array.reshape(length_second, 1)))

#print(construct_matrix(np.array([1, 2, 3]), np.array([4, 5, 6])))
```
